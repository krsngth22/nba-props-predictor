```python
from nba_api.stats.endpoints import playergamelog, commonallplayers, leaguegamefinder
from nba_api.stats.static import players, teams
import pandas as pd
import time
```

```python
all_players = players.get_active_players()
players_df = pd.DataFrame(all_players)
print(f"Total active players: {len(players_df)}")
players_df.head(10)
```

```python
def find_player(name):
    result = [p for p in all_players if name.lower() in p['full_name'].lower()]
    return result

find_player("LeBron")
```

```python
lebron_id = 2544

gamelog = playergamelog.PlayerGameLog(
    player_id=lebron_id,
    season='2024-25'
)

lebron_df = gamelog.get_data_frames()[0]
print(f"Games found: {len(lebron_df)}")
lebron_df.head()
```

```python
print(lebron_df.columns.tolist())
```

```python
time.sleep(1)

curry_id = 201939

curry_log = playergamelog.PlayerGameLog(
    player_id=curry_id,
    season='2024-25'
)

curry_df = curry_log.get_data_frames()[0]
print(f"Curry games found: {len(curry_df)}")
curry_df[['GAME_DATE','MATCHUP','PTS','REB','AST','MIN']].head(10)
```

```python
for name, df in [("LeBron", lebron_df), ("Curry", curry_df)]:
    print(f"\n{name}:")
    print(f"  Avg points:  {df['PTS'].mean():.1f}")
    print(f"  Avg rebounds: {df['REB'].mean():.1f}")
    print(f"  Avg assists:  {df['AST'].mean():.1f}")
    print(f"  Games played: {len(df)}")
```
